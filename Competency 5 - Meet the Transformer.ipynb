I installed the packages needed to run the transformer model.

```python
import torch
import transformers
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "distilbert/distilgpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

model.eval()

print("Model:", model_name)
print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
```

I loaded DistilGPT2, a pretrained AI model. It can generate text without needing to be trained again.

```python
examples = [
    "Hello world!",
    "Artificial intelligence is amazing.",
    "Unbelievable!"
]

for text in examples:
    ids = tokenizer.encode(text)
    tokens = tokenizer.convert_ids_to_tokens(ids)

    print("Text:", text)
    print("Tokens:", tokens)
    print("Token IDs:", ids)
    print("Decoded:", tokenizer.decode(ids))
    print()
```

I learned that AI breaks text into tokens. The word "Unbelievable!" was split into 5 tokens. Each token has its own ID, but they can be decoded back into the original text.

```python
prompt = "The future of artificial intelligence is"

inputs = tokenizer(prompt, return_tensors="pt")

for seed in [10, 20, 30]:
    torch.manual_seed(seed)

    output = model.generate(
        **inputs,
        max_new_tokens=30,
        do_sample=True,
        temperature=0.7,
        pad_token_id=tokenizer.eos_token_id
    )

    print("Seed:", seed)
    print(tokenizer.decode(output[0], skip_special_tokens=True))
    print()
```

I tested the same prompt with three different seeds. Each response was different. Some answers made more sense than others, showing that AI generation can be unpredictable.

```python
prompt = "The future of artificial intelligence is"
inputs = tokenizer(prompt, return_tensors="pt")

for temp in [0.3, 0.7, 1.2]:
    torch.manual_seed(42)

    output = model.generate(
        **inputs,
        max_new_tokens=30,
        do_sample=True,
        temperature=temp,
        pad_token_id=tokenizer.eos_token_id
    )

    print("Temperature:", temp)
    print(tokenizer.decode(output[0], skip_special_tokens=True))
    print()
```

I noticed that lower temperatures gave more predictable answers. Higher temperatures made the responses more random and less reliable.

```python
prompts = [
    "Explain artificial intelligence.",
    "Explain artificial intelligence to a 10-year-old.",
    "Explain artificial intelligence to a computer scientist."
]

for prompt in prompts:
    torch.manual_seed(42)

    inputs = tokenizer(prompt, return_tensors="pt")

    output = model.generate(
        **inputs,
        max_new_tokens=40,
        do_sample=True,
        temperature=0.7,
        pad_token_id=tokenizer.eos_token_id
    )

    print("Prompt:", prompt)
    print("Response:", tokenizer.decode(output[0], skip_special_tokens=True))
    print()
```

I changed the prompts to see how context affects the results. The responses changed, but the model did not always follow the instructions correctly.

```python
prompt = "The capital of France is"

torch.manual_seed(42)

inputs = tokenizer(prompt, return_tensors="pt")

output = model.generate(
    **inputs,
    max_new_tokens=40,
    do_sample=True,
    temperature=1.2,
    pad_token_id=tokenizer.eos_token_id
)

print(tokenizer.decode(output[0], skip_special_tokens=True))
```

I asked the model about the capital of France, but it never said Paris. Instead, it talked about military and economic security. This shows that the model can give confusing answers instead of useful information.

I learned that transformers use tokens to process and generate text. Training teaches a model patterns, while inference uses what it already learned. Changing the temperature, random seed, and prompt wording changed the results. DistilGPT2 was useful for testing, but I would not use it for customers without more testing because it does not always give accurate answers.
